```python
import numpy as np

def objective_function(x):
    # Sphere Function (Global Minimum is at x = [0, 0, ..., 0] with fitness 0)
    return np.sum(x**2)

def grey_wolf_optimizer(obj_func, dim, pack_size, max_iter, lb, ub):
    # Initialize alpha, beta, and delta positions and their fitness values
    Alpha_pos = np.zeros(dim)
    Alpha_score = float("inf")

    Beta_pos = np.zeros(dim)
    Beta_score = float("inf")

    Delta_pos = np.zeros(dim)
    Delta_score = float("inf")

    # Initialize the positions of search agents (pack size)
    Positions = np.random.uniform(0, 1, (pack_size, dim)) * (ub - lb) + lb

    # Convergence curve
    convergence_history = []

    # Main loop
    for l in range(0, max_iter):
        for i in range(0, pack_size):
            # Return back the search agents that go beyond the boundaries of the search space
            Positions[i, :] = np.clip(Positions[i, :], lb, ub)

            # Calculate objective function for each search agent
            fitness = obj_func(Positions[i, :])

            # Update Alpha, Beta, and Delta
            if fitness < Alpha_score:
                Alpha_score = fitness
                Alpha_pos = Positions[i, :].copy()

            if fitness > Alpha_score and fitness < Beta_score:
                Beta_score = fitness
                Beta_pos = Positions[i, :].copy()

            if fitness > Alpha_score and fitness > Beta_score and fitness < Delta_score:
                Delta_score = fitness
                Delta_pos = Positions[i, :].copy()

        # a decreases linearly from 2 to 0
        a = 2 - l * (2 / max_iter)

        # Update the Position of search agents including omegas
        for i in range(0, pack_size):
            for j in range(0, dim):

                # Update Alpha position equations
                r1 = np.random.random()
                r2 = np.random.random()
                A1 = 2 * a * r1 - a
                C1 = 2 * r2
                D_alpha = abs(C1 * Alpha_pos[j] - Positions[i, j])
                X1 = Alpha_pos[j] - A1 * D_alpha

                # Update Beta position equations
                r1 = np.random.random()
                r2 = np.random.random()
                A2 = 2 * a * r1 - a
                C2 = 2 * r2
                D_beta = abs(C2 * Beta_pos[j] - Positions[i, j])
                X2 = Beta_pos[j] - A2 * D_beta

                # Update Delta position equations
                r1 = np.random.random()
                r2 = np.random.random()
                A3 = 2 * a * r1 - a
                C3 = 2 * r2
                D_delta = abs(C3 * Delta_pos[j] - Positions[i, j])
                X3 = Delta_pos[j] - A3 * D_delta

                # Position update vector equation
                Positions[i, j] = (X1 + X2 + X3) / 3

        convergence_history.append(Alpha_score)
        if (l + 1) % 10 == 0 or l == 0:
            print(f"Iteration {l + 1}/{max_iter} - Best Fitness (Alpha): {Alpha_score:.6f}")

    return Alpha_pos, Alpha_score, convergence_history

# Run the optimizer
dim = 5           # Number of decision variables
pack_size = 10    # Number of wolves
max_iter = 50     # Number of iterations
lb = -10          # Lower bound
ub = 10           # Upper bound

best_position, best_fitness, history = grey_wolf_optimizer(objective_function, dim, pack_size, max_iter, lb, ub)
print(f"\nOptimal Solution Found: {best_position}")
print(f"Optimal Objective Value: {best_fitness:.6e}")
```
