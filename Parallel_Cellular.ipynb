```python
import numpy as np

# 1. Define the Problem: Minimize f(x) = x^2 - 4x + 4
def fitness_function(x):
    return x**2 - 4*x + 4

# 2. Initialize Parameters
rows, cols = 10, 10          # 100 cells in a 10x10 grid
iterations = 100             # Stopping criteria
lower_bound, upper_bound = -10, 10

# 3. Initialize Population randomly in [-10, 10]
grid = np.random.uniform(lower_bound, upper_bound, (rows, cols))

# Track best solution
best_solution = None
best_fitness = float('inf')

# 4 & 5. Evaluate Fitness and Iterate
for _ in range(iterations):
    # Evaluate fitness for all cells simultaneously
    fitness_grid = fitness_function(grid)

    # Update global best
    min_idx = np.unravel_index(np.argmin(fitness_grid), fitness_grid.shape)
    if fitness_grid[min_idx] < best_fitness:
        best_fitness = fitness_grid[min_idx]
        best_solution = grid[min_idx]

    new_grid = np.copy(grid)

    # Update States based on 3x3 neighborhood
    for i in range(rows):
        for j in range(cols):
            # Extract 3x3 neighborhood (with boundary clipping)
            r_min, r_max = max(0, i - 1), min(rows, i + 2)
            c_min, c_max = max(0, j - 1), min(cols, j + 2)

            neighborhood = grid[r_min:r_max, c_min:c_max]
            neighbor_fitness = fitness_grid[r_min:r_max, c_min:c_max]

            # Find best neighbor in the 3x3 window
            best_n_idx = np.unravel_index(np.argmin(neighbor_fitness), neighbor_fitness.shape)
            best_neighbor_val = neighborhood[best_n_idx]

            # Update Rule: Average of current/best neighbor (or neighborhood average)
            new_grid[i, j] = (grid[i, j] + best_neighbor_val) / 2.0

    grid = new_grid

# 6. Output Best Solution
print(f"Optimal x: {best_solution:.4f}, Minimum f(x): {best_fitness:.4f}")
```
